# Uniform Cost Search: Self-Driving Trailer from Nairobi to Kisumu

## 1. Introduction

This practical exercise implements Uniform Cost Search (UCS) to determine the optimal route for a self-driving trailer travelling from Nairobi to Kisumu through Kenya's highway network.

The directed graph and road distances used in this implementation are based on the group's Practical Exercise 3, Part 1 submission.

The search algorithm uses the distance between towns in kilometres as the path cost. It explores available routes by prioritising the path with the lowest accumulated cost until the destination, Kisumu, is reached.

The objective is to identify the optimal route and calculate its total travel distance.

```python

# Import the library needed for Uniform Cost Search
import heapq

print("Python library imported successfully!")
```

```python

# Kenya Highway Network
# Each town is connected to its neighbouring towns.
# The numbers represent road distances in kilometres.

graph = {
    'Nairobi': [('Thika', 42), ('Naivasha', 92), ('Narok', 142)],

    'Thika': [('Nairobi', 42), ("Murang'a", 40)],

    "Murang'a": [('Thika', 40), ('Nyeri', 60)],

    'Nyeri': [("Murang'a", 60), ('Nyahururu', 65)],

    'Nyahururu': [('Nyeri', 65), ('Nakuru', 90)],

    'Naivasha': [('Nairobi', 92), ('Nakuru', 65)],

    'Narok': [('Nairobi', 142), ('Bomet', 85)],

    'Bomet': [('Narok', 85), ('Kericho', 55), ('Kisii', 80)],

    'Nakuru': [('Naivasha', 65), ('Nyahururu', 90),
               ('Kericho', 100), ('Eldoret', 155)],

    'Kericho': [('Bomet', 55), ('Nakuru', 100), ('Kisumu', 90)],

    'Eldoret': [('Nakuru', 155), ('Kitale', 70), ('Kakamega', 90)],

    'Kitale': [('Eldoret', 70)],

    'Kisii': [('Bomet', 80), ('Kisumu', 110)],

    'Kakamega': [('Eldoret', 90), ('Kisumu', 55)],

    'Kisumu': [('Kericho', 90), ('Kisii', 110),
               ('Kakamega', 55), ('Busia', 100)],

    'Busia': [('Kisumu', 100)]
}

print("Kenya highway graph created successfully!")
print("Number of towns:", len(graph))
```

```python

# Uniform Cost Search Algorithm

import heapq
from itertools import count
from math import inf

def uniform_cost_search(graph, start, goal):

    # Priority queue stores: (total_cost, order, current_town, path)
    counter = count()
    priority_queue = [(0, next(counter), start, [start])]

    # Keep track of the cheapest known cost to each town
    best_cost = {start: 0}

    # Record the towns expanded during the search
    expanded_nodes = []

    while priority_queue:

        # Remove the route with the lowest total cost
        current_cost, _, current_town, path = heapq.heappop(priority_queue)

        # Ignore outdated routes
        if current_cost > best_cost.get(current_town, inf):
            continue

        # Record the current town and its accumulated cost
        expanded_nodes.append((current_town, current_cost))

        # Check whether the destination has been reached
        if current_town == goal:
            return path, current_cost, expanded_nodes

        # Explore neighbouring towns
        for neighbour, distance in graph[current_town]:

            new_cost = current_cost + distance

            # Only consider a route if it is cheaper
            if new_cost < best_cost.get(neighbour, inf):

                best_cost[neighbour] = new_cost

                new_path = path + [neighbour]

                heapq.heappush(
                    priority_queue,
                    (new_cost, next(counter), neighbour, new_path)
                )

    # Return this if no route exists
    return None, inf, expanded_nodes


print("Uniform Cost Search algorithm defined successfully!")
```

```python

# Run Uniform Cost Search from Nairobi to Kisumu

start = 'Nairobi'
goal = 'Kisumu'

optimal_path, total_cost, expanded_nodes = uniform_cost_search(
    graph, start, goal
)

print("UNIFORM COST SEARCH RESULTS")
print("---------------------------")

print("Starting Town:", start)
print("Destination:", goal)

print("\nOptimal Path:")
print(" -> ".join(optimal_path))

print("\nTotal Distance:", total_cost, "km")

print("\nExpanded Nodes:")
for town, cost in expanded_nodes:
    print(town, "-", cost, "km")
```

```python

# Display the Uniform Cost Search traversal

print("UNIFORM COST SEARCH - EXPANSION TRACE")
print("-------------------------------------")

for step, (town, cost) in enumerate(expanded_nodes, start=1):
    print(f"Step {step}: Expand {town} (Total cost: {cost} km)")

print("\nSearch completed when Kisumu was reached.")
```

## 2. Results and Discussion

The Uniform Cost Search algorithm successfully found the optimal route for the self-driving trailer travelling from Nairobi to Kisumu.

The optimal route identified was:

**Nairobi → Naivasha → Nakuru → Kericho → Kisumu**

The total path cost was calculated as:

92 + 65 + 100 + 90 = **347 km**

The algorithm explored alternative routes by prioritising the route with the lowest accumulated distance. It stopped when Kisumu was reached at the lowest available path cost.
